```python
def avg(arg) :
    sum = 0
    k = 0
    for i in arg :
        sum += i
        k += 1
    return sum/k
a=[1,2,3,4,5]
print(avg(a))
```

```python
import numpy as np
import scipy as stats
from scipy import stats
data=[10,20,25,42,30,42,45,50]

data_mean = np.mean(data)
print("mean:", data_mean)

data_median = np.median(data)
print("median:", data_median)

data_mode = stats.mode(data)
print("mode:", data_mode[0],data_mode.count)
```

```python
data_range=max(data)-min(data)
print("range:", data_range)
```

```python
data_var=np.var(data)
print("variance:", data_var)
```

```python
data1=[4,8]
variance = np.var(data1)
print("variance of data1:", variance)
```

```python
std_dev = np.std(data1)
print("standard deviation of data1:", std_dev)
```

```python
#IQR
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 75)
IQR = q2 - q1 #75th percentile - 25th percentile
print("25th percentile:", q1)
print("75th percentile:", q2)
print("IQR 75th percentile - 25th percentile:", IQR)
```

```python
# Lower Bound : Q1 - 1.5* IQR
# Upper Bound : Q2 + 1.5* IQR

LB = q1 - 1.5* IQR
UB = q2 + 1.5* IQR
print("Lower bound:", LB)
print("Upper bound:", UB)
```

```python
# Outliers
outlier = [x for x in data if x < LB or x > UB]
print("Outliers:", outlier)

```
