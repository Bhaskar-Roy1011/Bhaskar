**List**
1. List is used to store multiple items in.
2. List is ordered.
3. It allows duplicate value1s
4. It is mutable(Changeable)
5. It is defined by[].

```python
fruit = ['apple', 'mango',20,80,7,True,'apple',8j,'mango']
print(fruit)
```

```python
# Indexing
fruit = ['apple', 'mango',20,80,7,True,'apple',8j,'mango']
print(fruit[2])
```

```python
# Slicing
fruit = ['apple', 'mango',20,80,7,True,'apple',8j,'mango']
print(fruit[2:5])   # start: stop
```

```python
fruit = ['apple', 'mango',20,80,7,True,'apple',8j,'mango']
print(fruit[2:6:3])   # start:stop:step
```

```python
fruit = ['apple', 'mango',20,80,7,True,'apple',8j,'mango']
print(fruit[-5:-1])
```

```python
#Replaced
fruit = ['apple', 'mango',20,80,7,True,'apple',8j,'mango']
fruit[2] = 'banana'
print(fruit)
```

```python
#Append it is added in end of the list
fruit = ['apple', 'banana',20,80,7,True,'apple',8j,'mango']
fruit.append('kiwi')
print(fruit)   # for added the value at the end
```

```python
#Insert
fruit = ['apple', 'banana', 20, 80, 7, True, 'apple', 8j, 'mango', 'kiwi']
fruit.insert(2,'orange') # added the value in particular index
print(fruit)
```

```python
#Remove it remove the specific values from the index
fruit = ['apple', 'banana','orange', 20, 80, 7, True, 'apple', 8j, 'mango', 'kiwi']
fruit.remove('orange')
print(fruit)
```

```python
#Pop is takes the index values to be remove
fruit = ['apple', 'banana', 20, 80, 7, True, 'apple', 8j, 'mango', 'kiwi']
fruit.pop(2)
print(fruit)
```

```python
#Clear
fruit = ['apple', 'banana', 20, 80, 7, True, 'apple', 8j, 'mango', 'kiwi']
fruit.clear()
print(fruit)
```

```python
#Delete
del fruit
```
