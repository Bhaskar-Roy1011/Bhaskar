```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
```

**Steps:**
**a> Load the Iris Dataset**

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
column_names = ["sepal_length", "sepal_width", "petal_length", "petal_width", "class"]
iris_data = pd.read_csv(url, names=column_names)
print("Iris Dataset Loaded Successfully....")
iris_data.head(10)
```

**b> Exlore and visualize the data using Pandas and Mataplotlib/Seaborn**

```python
iris_data.iloc[50:70]
```

```python
iris_data.describe()
```

```python
sns.pairplot(iris_data, hue='class')
plt.show()
```

**c> Train a model using a classifier (KNN)** 

```python
x = iris_data.drop('class', axis=1)
y = iris_data['class']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train, y_train)
```

**d> Evaluate model accuracy using confusion matrix and classification report**

```python
y_pred = knn.predict(x_test)
accuracy = accuracy_score(y_test,y_pred)
print("Accuracy:",accuracy)
```

```python
print(classification_report(y_test, y_pred))
```
